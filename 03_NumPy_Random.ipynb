# NumPy Random Module

## Introduction

The NumPy random module provides functions for generating random numbers, selecting random elements, rearranging data, and generating values from probability distributions.

Random operations are useful in simulations, statistical experiments, data sampling, and Machine Learning.

### Why do we use the NumPy random module?

- To generate random numbers for experiments.
- To select random samples from datasets.
- To shuffle data before training Machine Learning models.
- To simulate real-world events.
- To generate random values from probability distributions.

### Real-world example

A teacher can randomly select students for a classroom activity.

### Business example

A company can randomly select customer records for a survey.

### AI/ML example

Random numbers are used in model initialization, data shuffling, and simulation-based experiments.

## Importing NumPy

We import NumPy using the alias `np` so that its functions can be accessed easily.

```python
import numpy as np

print("NumPy version:", np.__version__)
```

## 1. Generating Random Numbers

Random numbers are values generated through a process designed to produce unpredictable-looking results. NumPy provides random number generators to generate values for numerical computations and simulations.

The modern NumPy approach uses `np.random.default_rng()` to create a random number generator.

The `random()` method generates floating-point numbers in the interval from 0 (inclusive) to 1 (exclusive).

### Real-world example
A game can generate a random number to decide a player's score bonus.

### Business example
A business can use random values to simulate variations in customer demand.

### AI/ML use case
Random numbers can help initialize model parameters and perform simulations.

### Python implementation

```python
import numpy as np

rng = np.random.default_rng()

random_values = rng.random(5)

print("Random values:", random_values)
print("Minimum value:", random_values.min())
print("Maximum value:", random_values.max())
```

## 2. Random Seed

A random seed initializes a pseudo-random number generator. Using the same seed with the same generator and sequence of operations makes the generated results reproducible.

This is important when testing code, comparing experiments, and debugging Machine Learning workflows.

### Real-world example
A teacher can use a fixed seed to generate the same randomized question order when demonstrating a program.

### Business example
An analyst can reproduce the same randomly selected sample while checking a report.

### AI/ML use case
A fixed seed helps reproduce data sampling and randomized experiments. However, setting a seed alone does not guarantee identical results across every library version, hardware configuration, or parallel execution environment.

### Python implementation

```python
rng1 = np.random.default_rng(42)
rng2 = np.random.default_rng(42)

values1 = rng1.random(5)
values2 = rng2.random(5)

print("First sequence:", values1)
print("Second sequence:", values2)
print("Sequences match:", np.array_equal(values1, values2))
```

## 3. Generating Random Integers

The `integers()` method generates random integer values within a specified range.

By default, the lower limit is included and the upper limit is excluded. The `endpoint=True` option can include the upper limit.

### Real-world example
A dice simulation can generate a random integer from 1 to 6.

### Business example
A retailer can simulate a random number of customer visits for a simple experiment.

### AI/ML use case
Random integers can be used to select indices, generate synthetic test data, or sample records.

### Python implementation

```python
rng = np.random.default_rng(42)

dice_rolls = rng.integers(1, 7, size=10)

print("Dice rolls:", dice_rolls)
print("Minimum roll:", dice_rolls.min())
print("Maximum roll:", dice_rolls.max())
```

## 4. Random Choice

The `choice()` method selects random elements from a given array or sequence.

The `size` parameter specifies how many elements to select. By default, sampling is performed with replacement, meaning an element may be selected more than once. Setting `replace=False` selects without replacement.

Optional probabilities can be supplied using the `p` parameter, provided they are valid and sum to 1.

### Real-world example
A teacher can randomly select three students for a presentation.

### Business example
A company can select customers to participate in a feedback survey.

### AI/ML use case
Random choice can help create samples from a dataset for testing or analysis.

### Python implementation

```python
rng = np.random.default_rng(42)

students = np.array(["Asha", "Ravi", "Meena", "Kiran", "Pooja"])

selected_students = rng.choice(students, size=3, replace=False)

print("All students:", students)
print("Selected students:", selected_students)
```

## 5. Shuffle

The `shuffle()` method randomly rearranges the elements of an array in place. This means it modifies the original array.

For a multidimensional array, `shuffle()` rearranges elements along the specified axis; by default, it operates along axis 0.

### Real-world example
A teacher can shuffle the order of students before assigning presentation turns.

### Business example
A company can shuffle customer records before running a sampling experiment.

### AI/ML use case
Shuffling training data can help prevent a model from learning unwanted patterns from the original order of records.

### Python implementation

```python
rng = np.random.default_rng(42)

original = np.array([10, 20, 30, 40, 50])
data = original.copy()

rng.shuffle(data)

print("Original array:", original)
print("Shuffled array:", data)
```

## 6. Permutation

The `permutation()` method returns a randomly rearranged sequence or array without modifying the original array.

When given an integer `n`, it returns a permutation of the integers from `0` to `n-1`.

### Difference between shuffle and permutation

- `shuffle()` changes the given array in place.
- `permutation()` returns a rearranged result and leaves the original array unchanged.

### Real-world example
A school can generate a randomized order of student roll numbers.

### AI/ML use case
Permutations can be used to randomize record order before training or evaluating a model.

### Python implementation

```python
rng = np.random.default_rng(42)

numbers = np.array([10, 20, 30, 40, 50])
permuted = rng.permutation(numbers)

print("Original array:", numbers)
print("Permuted array:", permuted)
```

## 7. Random Distributions

A probability distribution describes how possible values are distributed and how likely those values are under a specified model.

NumPy provides methods for generating random samples from many distributions.

Common examples include:

- **Uniform distribution:** Values within a specified interval have equal probability density.
- **Normal distribution:** Values cluster around a mean and follow a bell-shaped distribution.
- **Binomial distribution:** Represents the number of successes in a fixed number of independent trials with the same success probability.

### Real-world example
A normal distribution can be used as a simplified model of measurement errors.

### Business example
A uniform distribution can simulate demand values when every value in a range is assumed equally likely.

### AI/ML use case
Distributions are useful for simulations, generating synthetic data, and understanding randomness in statistical and Machine Learning experiments.


### 8.1 Uniform distribution

```python
rng = np.random.default_rng(42)

uniform_values = rng.uniform(low=10, high=20, size=5)

print("Uniform samples:", uniform_values)
print("Minimum:", uniform_values.min())
print("Maximum:", uniform_values.max())
```

### 8.2 Normal distribution

```python
rng = np.random.default_rng(42)

normal_values = rng.normal(loc=50, scale=10, size=1000)

print("Sample mean:", round(normal_values.mean(), 2))
print("Sample standard deviation:", round(normal_values.std(), 2))
print("First five values:", normal_values[:5])
```

#### Output explanation:
- loc=50 specifies the distribution mean.
- scale=10 specifies the standard deviation.
- size=1000 generates 1,000 samples.

### 8.3 Binomial distribution

```python
rng = np.random.default_rng(42)

successes = rng.binomial(n=10, p=0.5, size=10)

print("Success counts:", successes)
print("Minimum possible successes:", successes.min())
print("Maximum possible successes:", successes.max())
```

#### Output explanation:
Each generated value represents the number of successes in 10 independent trials, where the probability of success on each trial is 0.5. Each count must be between 0 and 10 inclusive.
