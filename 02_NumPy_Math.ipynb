# NumPy Mathematical Operations

## 1. Introduction

NumPy provides mathematical and statistical functions that help us perform calculations efficiently on numerical arrays. These functions are useful for analyzing data, summarizing information, and performing calculations required by Machine Learning algorithms.

## 2. Why do we use NumPy mathematical operations?

- To calculate totals and averages.
- To measure the spread of data.
- To identify minimum and maximum values.
- To perform matrix and vector calculations.
- To prepare numerical data for Machine Learning.

### Real-world example
A teacher can calculate the average, highest, and lowest marks of students.

### Business example
A company can calculate total sales, average revenue, and sales variation.

### AI/ML example
Machine Learning uses matrix multiplication, statistical calculations, and linear algebra to process features and make predictions.

## 3. Importing NumPy

We use the alias `np` to access NumPy functions conveniently.

```python
import numpy as np

print("NumPy version:", np.__version__)

marks = np.array([60, 70, 80, 90, 100])
print("Marks:", marks)
```

### output explanation:
the installed NumPy version and the array [60 70 80 90 100].

## 4. Sum

The `np.sum()` function calculates the total of all elements in an array. For multidimensional arrays, the `axis` parameter can calculate totals along a particular axis.

### Example
A teacher wants to find the total marks obtained by students.

### AI/ML use case
Sum is used to calculate totals, aggregate values, and combine numerical contributions in calculations.

```python
marks = np.array([60, 70, 80, 90, 100])

print("Total marks:", np.sum(marks))

sales = np.array([[100, 200, 300],
                  [150, 250, 350]])

print("Total sales:", np.sum(sales))
print("Sum by column:", np.sum(sales, axis=0))
print("Sum by row:", np.sum(sales, axis=1))
```

### Output explanation: 
np.sum() adds all elements. axis=0 gives column totals, while axis=1 gives row totals.

## 5. Mean

The mean is the arithmetic average of a collection of numbers. It is calculated by dividing the sum of all values by the number of values.

Formula:

Mean = Sum of all values / Number of values

### Example
The mean of marks helps summarize a student's average performance.

### AI/ML use case
The mean is used in data analysis, feature scaling, and handling missing numerical values when appropriate.

```python
marks = np.array([60, 70, 80, 90, 100])

print("Mean marks:", np.mean(marks))
```

### Output explanation: 
The total is 400 and there are five values, so the mean is 80.

## 6. Median

The median is the middle value after the data is arranged in sorted order. When the number of observations is even, it is the average of the two middle values.

The median is less affected by extreme values than the mean.

### Business example
Median salary can help summarize typical earnings when a few salaries are unusually high.

### AI/ML use case
The median can be used to fill missing numerical values when the data distribution makes it an appropriate choice.

```python
values = np.array([10, 20, 30, 40, 1000])

print("Median:", np.median(values))
print("Mean:", np.mean(values))
```

### Output explanation:
The extreme value 1000 pulls the mean upward, while the median remains 30.

## 7. Variance

Variance measures how far data values are spread out from their mean. A low variance means the values are closer to the mean, while a high variance means they are more spread out.

For population variance, NumPy calculates the average of the squared differences from the mean using `np.var()` by default (`ddof=0`).

### AI/ML use case
Variance helps understand feature variability and is important in statistical analysis and some feature-selection techniques.

```python
values = np.array([10, 20, 30, 40, 50])

print("Mean:", np.mean(values))
print("Population variance:", np.var(values))
print("Sample variance:", np.var(values, ddof=1))
```

### Output explanation: 
The population variance is 200. Setting ddof=1 calculates sample variance by dividing the squared deviations by \(n-1\), rather than \(n\).

## 8. Standard Deviation

Standard deviation measures the spread of data around its mean. It is the square root of variance and is expressed in the same units as the original data.

NumPy's `np.std()` calculates population standard deviation by default. Use `ddof=1` for sample standard deviation.

### Business example
A business can compare the variation in daily sales to understand how consistent its sales are.

### AI/ML use case
Standard deviation is used in standardization, where features are often centered around their mean and scaled by their standard deviation.

```python
values = np.array([10, 20, 30, 40, 50])

print("Population standard deviation:", np.std(values))
print("Sample standard deviation:", np.std(values, ddof=1))
```

### Output explanation:
Standard deviation is the square root of variance. The sample standard deviation is larger here because it uses \(n-1\) in the denominator.

## 9. Minimum and Maximum

`np.min()` returns the smallest value in an array, while `np.max()` returns the largest value.

These functions help identify the range of values in a dataset.

### Real-world example
A teacher can find the highest and lowest marks in a class.

### AI/ML use case
Minimum and maximum values are useful when examining feature ranges and preparing data for min-max scaling.

```python
temperatures = np.array([28, 32, 25, 35, 30])

print("Minimum temperature:", np.min(temperatures))
print("Maximum temperature:", np.max(temperatures))
print("Range:", np.max(temperatures) - np.min(temperatures))
```

### Output explanation: 
The smallest value is 25, the largest is 35, and their difference is 10.

## 10. Argmin and Argmax

`np.argmin()` returns the index of the minimum value, while `np.argmax()` returns the index of the maximum value.

For multidimensional arrays, these functions return flattened indices by default unless an axis is specified.

### Business example
A company can identify the position of the day with the lowest or highest sales.

### AI/ML use case
These functions are useful for finding the index of the highest prediction score or lowest error value.

```python
sales = np.array([500, 300, 700, 200, 900])

print("Index of minimum sales:", np.argmin(sales))
print("Index of maximum sales:", np.argmax(sales))
print("Minimum sales:", sales[np.argmin(sales)])
print("Maximum sales:", sales[np.argmax(sales)])
```

### Output explanation:
NumPy uses zero-based indexing. The minimum value, 200, is at index 3, and the maximum value, 900, is at index 4.

## 11. Percentile

A percentile indicates the value below which a given percentage of observations falls, according to the chosen percentile calculation method.

For example, the 50th percentile corresponds to the median. NumPy uses `np.percentile()` to calculate percentiles.

### Business example
A company can analyze the 90th percentile of delivery times to understand a high-end delivery duration.

### AI/ML use case
Percentiles are useful for understanding data distributions, setting thresholds, and identifying potential outliers.

```python
scores = np.array([10, 20, 30, 40, 50])

print("25th percentile:", np.percentile(scores, 25))
print("50th percentile:", np.percentile(scores, 50))
print("75th percentile:", np.percentile(scores, 75))
```

### Output explanation: 
These values use NumPy's default percentile method. Other datasets and percentile methods may produce different results.

## 12. Dot Product

The dot product multiplies corresponding elements of two vectors and adds the products together.

For vectors \(a\) and \(b\):

\[
a \cdot b = \sum_i a_i b_i
\]

NumPy provides `np.dot()` to calculate the dot product of two one-dimensional arrays.

### Real-world example
A shopping bill can be calculated by multiplying each product's quantity by its price and summing the results.

### AI/ML use case
Dot products are used in weighted sums, linear models, and neural network calculations.

```python
quantities = np.array([2, 3, 4])
prices = np.array([100, 50, 25])

total_cost = np.dot(quantities, prices)

print("Total cost:", total_cost)
```

### Output explanation:
The total is \(2 \times 100 + 3 \times 50 + 4 \times 25 = 450\).

## 13. Matrix Multiplication

Matrix multiplication combines rows of the first matrix with columns of the second matrix. The number of columns in the first matrix must equal the number of rows in the second matrix.

NumPy supports matrix multiplication using the `@` operator or `np.matmul()`.

### AI/ML use case
Matrix multiplication is a fundamental operation in neural networks, linear transformations, and many Machine Learning algorithms.

```python
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

result = A @ B

print("Matrix A:\n", A)
print("Matrix B:\n", B)
print("Matrix multiplication:\n", result)
```

### Output explanation:
The top-left result is \(1 \times 5 + 2 \times 7 = 19\). The remaining elements are calculated using the same row-by-column method.
### Important difference:
A * B performs element-wise multiplication, while A @ B performs matrix multiplication.

## 14. Transpose of a Matrix

The transpose of a matrix exchanges its rows and columns. NumPy provides the `.T` attribute and `np.transpose()` function.

For example, a matrix with shape `(2, 3)` becomes a matrix with shape `(3, 2)` after transposition.

### AI/ML use case
Transposition is used to rearrange data dimensions and perform linear algebra calculations.

```python
matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])

transposed = matrix.T

print("Original matrix:\n", matrix)
print("Original shape:", matrix.shape)
print("Transposed matrix:\n", transposed)
print("Transposed shape:", transposed.shape)
```

### Output explanation: 
The original two-row, three-column matrix becomes a three-row, two-column matrix.

## 15. Basic Linear Algebra Operations

Linear algebra deals with vectors, matrices, and operations involving them. NumPy provides functions for solving systems of equations, calculating determinants, finding inverses, and computing other matrix operations.

Important functions include:

- `np.linalg.solve()` – solves a system of linear equations.
- `np.linalg.det()` – calculates a square matrix's determinant.
- `np.linalg.inv()` – calculates the inverse of a nonsingular square matrix.
- `np.linalg.norm()` – calculates a vector or matrix norm.

### AI/ML use case
Linear algebra is used in regression, transformations, optimization, and neural network computations.

### Important note
Not every matrix has an inverse. For solving linear equations, `np.linalg.solve()` is generally preferable to explicitly calculating an inverse.
